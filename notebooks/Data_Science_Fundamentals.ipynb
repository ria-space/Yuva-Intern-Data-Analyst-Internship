**Practical component: Student Performance Data Analysis**

Dataset is simulated for educational purposes and does not represent real students.

## 1. Import libraries

Install missing packages if needed: `pip install pandas numpy matplotlib seaborn`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
```

## 2. Load and inspect the dataset

```python
from pathlib import Path

# Run from the assessment folder or update this relative path if needed.
data_path = Path("../dataset/student_performance.csv")
df = pd.read_csv(data_path)

print("Shape:", df.shape)
display(df.head())
df.info()
```

## 3. Data quality checks

```python
print("Missing values by column:")
display(df.isna().sum())

print("Duplicate rows:", df.duplicated().sum())
print("Duplicate student IDs:", df["student_id"].duplicated().sum())

print("Age range:", df["age"].min(), "to", df["age"].max())
print("Attendance range:", df["attendance_percentage"].min(), "to", df["attendance_percentage"].max())
print("Marks range:", df["final_exam_marks"].min(), "to", df["final_exam_marks"].max())
```

## 4. Data types and descriptive statistics

```python
print(df.dtypes)
display(df.describe())

marks = df["final_exam_marks"]
print("Mean:", marks.mean())
print("Median:", marks.median())
print("Mode(s):", marks.mode().tolist())
print("Range:", marks.max() - marks.min())
print("Sample variance:", marks.var(ddof=1))
print("Sample standard deviation:", marks.std(ddof=1))
```

## 5. Grouped analysis

```python
display(
    df.groupby("gender")["final_exam_marks"]
      .agg(["count", "mean", "median", "std"])
      .round(2)
)
```

## 6. Correlation analysis

Correlation measures linear association, not causation.

```python
correlation = df["study_hours_per_day"].corr(df["final_exam_marks"])
print("Pearson correlation (study hours vs final marks):", round(correlation, 3))
display(df[["study_hours_per_day", "attendance_percentage",
            "previous_exam_marks", "final_exam_marks"]].corr().round(2))
```

## 7. Visualizations

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["final_exam_marks"], bins=8, kde=True)
plt.title("Distribution of Final Exam Marks")
plt.xlabel("Final exam marks")
plt.ylabel("Number of students")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="study_hours_per_day", y="final_exam_marks", hue="gender")
plt.title("Study Hours and Final Exam Marks")
plt.xlabel("Study hours per day")
plt.ylabel("Final exam marks")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="gender", y="final_exam_marks")
plt.title("Final Exam Marks by Gender")
plt.xlabel("Gender category")
plt.ylabel("Final exam marks")
plt.tight_layout()
plt.show()
```

## 8. Key observations and limitations

- The dataset contains 20 simulated records and 8 columns.
- Review the computed descriptive statistics to understand the distribution of final marks.
- The scatter plot and Pearson coefficient describe the linear association between study hours and final marks in this constructed dataset.
- The dataset is small and synthetic; findings cannot be generalized to real students.
- Correlation does not prove that study hours cause changes in marks. Other variables may influence performance.

This notebook demonstrates loading data, inspecting structure, checking quality, calculating descriptive statistics, examining correlation and communicating patterns through plots. The results are illustrative and should be interpreted within the limitations of the simulated dataset.
