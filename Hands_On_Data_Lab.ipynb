# Assessment 02: Hands-On Data Lab Implementation

This notebook demonstrates environment use, data inspection, cleaning, manipulation, visualization, and basic analysis. The included 20-row dataset is synthetic and for practice only.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/sample_sales_data.csv', parse_dates=['date'])
df.head()
```

## 1. Inspect the dataset

```python
print('Shape:', df.shape)
print('\nData types:\n', df.dtypes)
print('\nDescriptive statistics:\n', df.describe(include='all'))
```

## 2. Data quality checks and cleaning

```python
print('Missing values:\n', df.isna().sum())
print('Duplicate rows:', df.duplicated().sum())
df = df.drop_duplicates().dropna(subset=['order_id','date','category','quantity','unit_price'])
df = df[(df.quantity > 0) & (df.unit_price >= 0)].copy()
df['sales'] = df.quantity * df.unit_price
df.head()
```

## 3. Data manipulation

```python
high_value = df[df.sales >= 5000].sort_values('sales', ascending=False)
sales_by_category = df.groupby('category', as_index=False).sales.sum().sort_values('sales', ascending=False)
sales_by_city = df.groupby('city', as_index=False).agg(total_sales=('sales','sum'), orders=('order_id','count'))
print(high_value)
print(sales_by_category)
print(sales_by_city)
```

## 4. Basic analysis

```python
print('Total sales:', df.sales.sum())
print('Average order value:', round(df.sales.mean(), 2))
print('Total units:', df.quantity.sum())
print('Average rating:', round(df.rating.mean(), 2))
print(df[['quantity','unit_price','sales','rating']].corr(numeric_only=True))
```

## 5. Visualizations

```python
sns.barplot(data=sales_by_category, x='category', y='sales')
plt.title('Total Sales by Category'); plt.xticks(rotation=25); plt.tight_layout(); plt.show()
```

```python
daily = df.groupby('date', as_index=False).sales.sum()
sns.lineplot(data=daily, x='date', y='sales', marker='o')
plt.title('Daily Sales Trend'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

```python
sns.scatterplot(data=df, x='quantity', y='sales', hue='category')
plt.title('Quantity vs Sales'); plt.tight_layout(); plt.show()
```

## 6. Export and limitations
The cleaned dataset is exported to `outputs/cleaned_sales_data.csv`. This small synthetic dataset demonstrates methods only; it cannot support real business conclusions.

```python
os.makedirs('../outputs', exist_ok=True)
df.to_csv('../outputs/cleaned_sales_data.csv', index=False)
print('Export complete.')
```
