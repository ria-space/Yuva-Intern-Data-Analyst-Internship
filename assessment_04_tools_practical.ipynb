# Assessment 04 – Data Science Tools and Libraries\n\nPractical exercises using an illustrative UPI transaction dataset. The larger associated portfolio project uses a synthetic dataset.

## 1. Import libraries

```python
import numpy as np\nimport pandas as pd\nimport matplotlib.pyplot as plt\nimport seaborn as sns\n%matplotlib inline
```

## 2. Create an illustrative dataset\n\nThis sample makes the notebook independently runnable. For the full project, load `data/processed/upi_transactions_risk_scored.csv` instead.

```python
df = pd.DataFrame({
    "transaction_id":["TX001","TX002","TX003","TX004","TX005","TX006","TX007","TX008"],
    "amount_inr":[250,1200,5000,150,8200,600,3400,950],
    "transaction_type":["P2P","P2M","P2M","Recharge","Bill Payment","P2P","Bill Payment","P2M"],
    "transaction_status":["SUCCESS","SUCCESS","FAILED","SUCCESS","SUCCESS","FAILED","SUCCESS","SUCCESS"],
    "device_type":["Android","iOS","Web","Android","Web","iOS","Android","Android"],
    "fraud_flag":[0,0,1,0,1,0,0,0]
})
df.head()
```

## 3. Inspect structure and data quality

```python
print('Shape:', df.shape)\ndf.info()\ndisplay(df.describe(include='all'))\ndisplay(df.isna().sum())\nprint('Duplicate rows:', df.duplicated().sum())
```

## 4. Pandas filtering and aggregation

```python
high_value = df[df["amount_inr"] >= 1000]
display(high_value)
summary = df.groupby("transaction_type").agg(
    transaction_count=("transaction_id","count"),
    average_amount=("amount_inr","mean"),
    flagged_count=("fraud_flag","sum")
).reset_index()
display(summary.round(2))
```

## 5. NumPy numerical calculations

```python
amounts=df['amount_inr'].to_numpy()\nprint('Mean:',np.mean(amounts))\nprint('Median:',np.median(amounts))\nprint('Standard deviation:',np.std(amounts))\nprint('90th percentile:',np.percentile(amounts,90))
```

## 6. Visualizations with Matplotlib and Seaborn

```python
plt.figure(figsize=(8,4))
sns.countplot(data=df,x="transaction_type")
plt.title("Transaction Count by Type")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.histplot(data=df,x="amount_inr",bins=6,kde=True)
plt.title("Transaction Amount Distribution")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.boxplot(data=df,x="fraud_flag",y="amount_inr")
plt.title("Transaction Amount by Dataset Fraud Flag")
plt.xlabel("Fraud flag (1 = flagged)")
plt.tight_layout()
plt.show()
```

## 7. Derived metric and interpretation

```python
flagged_rate=df['fraud_flag'].mean()*100\nprint(f'Flagged transaction rate: {flagged_rate:.2f}%')\nprint('Average amount:',round(df['amount_inr'].mean(),2))\nprint('Maximum amount:',df['amount_inr'].max())
```
